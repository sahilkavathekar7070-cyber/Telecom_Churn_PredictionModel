```python
import pandas as pd
df= pd.read_csv(r"C:\Telecom_Churn_PredictionModel\data\raw\churn.csv")
```

```python
numerical_col=df.select_dtypes(exclude = 'object').columns
categorical_col = df.select_dtypes(include = 'object').columns

stats= []

from collections import OrderedDict
for i in numerical_col :
    numerical_stat = OrderedDict({
        'Feature' : i,
        'Count' : df[i].count(),
        'Missing Values': df[i].isnull().sum(),
        "%of Missing Values" : df[i].isnull().sum() / len(df) * 100,
        'Minimum' : df[i].min(),
        'Maximum' : df[i].max(),
        'Mean' : df[i].mean(),
        'Median' : df[i].median(),
    })
    stats.append(numerical_stat)
    report=pd.DataFrame(stats)
report
```
